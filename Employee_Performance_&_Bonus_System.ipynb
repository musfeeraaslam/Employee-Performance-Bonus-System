```python
def calculate_bonus(name,salary,projects,performance):
  if performance >= 90:
    performance_label = "Excellent"
  elif performance >= 80:
    performance_label = "Very Good"
  elif performance >= 70:
    performance_label = "Good"
  else:
    performance_label = "Needs Imporvement"

  if performance >=90 and projects >= 3:
    bonus_rate = 0.20
  elif performance >=80 and projects >= 3:
    bonus_rate = 0.15
  elif performance >=70 and projects >= 2:
    bonus_rate = 0.10
  else:
    bonus_rate = 0

  bonus = salary * bonus_rate
  final_salary = bonus + salary

  print(f"Employee: {name}")
  print(f"Performance: {performance_label}")
  print(f"Projects: {projects}")
  print(f"Bonus: {bonus:.0f}")
  print(f"Final Salary: {final_salary:.0f}")

name        = input("Enter employee name: ")
salary      = float(input("Enter salary: "))
projects    = int(input("Enter projects completed: "))
performance = int(input("Enter performance score (0-100): "))

calculate_bonus(name, salary, projects, performance)
```
